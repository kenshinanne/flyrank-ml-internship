```python
# Load the anonymized starter dataset.
# This works if the repo is available locally, or when opened in Colab.
from pathlib import Path
import pandas as pd

possible_paths = [
    Path("../../data/raw/content_refresh_anonymized.csv"),
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("/content/content_refresh_anonymized.csv"),
]
found = next((p for p in possible_paths if p.exists()), None)
if found:
    df = pd.read_csv(found)
else:
    # Requires an internet connection and access to the GitHub repository.
    url = "https://raw.githubusercontent.com/kenshinanne/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
    df = pd.read_csv(url)
print("Dataset loaded:", df.shape)
```

## 2. The question: decision, action, cost of a wrong call

**Research question:** Can we group website pages based on their performance patterns to identify which pages may need attention?

One row represents one content item (website page) in this starter dataset. I want to compare search and content signals to find groups with similar patterns. A content team could use these groups to decide which pages to **review, improve, or monitor**. If the groups are misleading, the team may spend time on the wrong pages or miss pages that need attention.

**Planned output:** A short description of each group and suggested review actions. I am not building a model yet in Week 1.

```python
# Check the unit of analysis and a few columns that could support grouping.
features_to_check = ["impressions_90d", "clicks_90d", "content_age_days", "avg_position", "engagement_rate"]
print("Unique content items:", df["content_id"].nunique())
print("Rows:", len(df))
print("Available signals:", [c for c in features_to_check if c in df.columns])
```
